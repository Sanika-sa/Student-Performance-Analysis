```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for cleaner visualizations
sns.set_theme(style="whitegrid")

# Create outputs/charts directory if it doesn't exist
os.makedirs("outputs/charts", exist_ok=True)

# Set up figure styling
plt.figure(figsize=(12, 10))

# 1. Distribution Plot: Target Score Distribution
plt.subplot(2, 2, 1)
# Automatically detect the target performance column
target_col = 'FinalScore' if 'FinalScore' in df.columns else df.select_dtypes(include='number').columns[-1]
sns.histplot(df[target_col], kde=True, color='blue', bins=15)
plt.title(f"Distribution of {target_col}")
plt.xlabel(target_col)
plt.ylabel("Frequency")

# 2. Comparison Plot: Box Plot (Categorical vs Numerical)
plt.subplot(2, 2, 2)
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
cat_col = categorical_cols[0] if len(categorical_cols) > 0 else df.columns[1]
sns.boxplot(x=cat_col, y=target_col, data=df, palette='Set2')
plt.title(f"{target_col} by {cat_col}")
plt.xlabel(cat_col)
plt.ylabel(target_col)
plt.xticks(rotation=45)

# 3. Relationship Plot: Scatter Plot (Study Hours vs Target Score)
plt.subplot(2, 2, 3)
numeric_cols = df.select_dtypes(include='number').columns
study_col = next((col for col in numeric_cols if 'study' in col.lower() or 'hour' in col.lower()), numeric_cols[0])
sns.scatterplot(x=study_col, y=target_col, data=df, alpha=0.7, color='purple')
plt.title(f"{study_col} vs {target_col}")
plt.xlabel(study_col)
plt.ylabel(target_col)

# 4. Correlation Heatmap
plt.subplot(2, 2, 4)
corr = df.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f', cbar=True, linewidths=0.5)
plt.title("Correlation Heatmap")

plt.tight_layout()
# Save the combined charts overview
plt.savefig("outputs/charts/comprehensive_analysis.png", dpi=300)
plt.show()

print("Visualizations generated and saved successfully in 'outputs/charts/'!")
```
